### 1. Vad kännetecknar klassificeringsproblem? Ge några exempel på tillämpningsområden.

Klassificeringsproblem kännetecknas av att utdatan (y) är diskret, alltså den tillhör en av ett bestämt antal kategorier, till skillnad från regression där y kan anta vilket värde som helst. Är det bara två kategorier kallas det binär klassificering (typ spam/inte spam), fler kategorier kallas multiklass (typ bildigenkänning av djur).

### 2. Förklara hur OvR- och OvO-algoritmerna fungerar.

OvR (One vs Rest) tränar en binär modell per klass, där den klassen ställs mot alla andra klasser slagna ihop. Har vi t.ex. 3 klasser blir det 3 modeller. OvO (One vs One) tränar istället en modell för varje par av klasser, så med 3 klasser blir det 3 modeller (A mot B, A mot C, B mot C), och med fler klasser blir det snabbt många fler modeller än OvR.

### 3. Förklara följande utvärderingsmått: a) Confusion matrix, b) Accuracy, c) Precision, d) Recall, e) F1-score, f) ROC-kurvan

a) Confusion matrix - en tabell som visar hur många rätt och fel modellen gjort, uppdelat per kategori (t.ex. hur många "ja" som gissades rätt/fel och hur många "nej" som gissades rätt/fel).

b) Accuracy - andelen rätta gissningar delat på totala antalet gissningar.

c) Precision - av alla gånger modellen gissade "ja", hur stor andel var faktiskt rätt.

d) Recall - av alla som faktiskt var "ja", hur många lyckades modellen hitta.

e) F1-score - ett balanserat värde mellan precision och recall, bra när man vill ha en helhetsbild istället för att bara kolla på ett av måtten.

f) ROC-kurvan - en graf som visar hur bra modellen är på att skilja mellan klasserna vid olika tröskelvärden.

### 4. Vad är precision-recall tradeoff för något?

Precision-recall tradeoff betyder att man måste välja mellan hög precision eller hög recall, man kan sällan få båda maximerade samtidigt. Höjer man tröskeln för att räknas som "ja" blir modellen mer försiktig, precisionen går upp men recall går ner (missar fler). Sänker man tröskeln fångar modellen fler "ja" (högre recall) men precisionen sjunker eftersom fler blir felgissade.
