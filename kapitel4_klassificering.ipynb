## Faktafrågor

### 1. Vad kännetecknar klassificeringsproblem? Ge några exempel på tillämpningsområden.

Klassificeringsproblem kännetecknas av att utdatan (y) är diskret, alltså den tillhör en av ett bestämt antal kategorier, till skillnad från regression där y kan anta vilket värde som helst. Är det bara två kategorier kallas det binär klassificering (typ spam/inte spam), fler kategorier kallas multiklass (typ bildigenkänning av djur).

### 2. Förklara hur OvR- och OvO-algoritmerna fungerar.

OvR (One vs Rest) tränar en binär modell per klass, där den klassen ställs mot alla andra klasser slagna ihop. Har vi t.ex. 3 klasser blir det 3 modeller. OvO (One vs One) tränar istället en modell för varje par av klasser, så med 3 klasser blir det 3 modeller (A mot B, A mot C, B mot C), och med fler klasser blir det snabbt många fler modeller än OvR.

### 3. Förklara följande utvärderingsmått: a) Confusion matrix, b) Accuracy, c) Precision, d) Recall, e) F1-score, f) ROC-kurvan

a) Confusion matrix - en tabell som visar hur många rätt och fel modellen gjort, uppdelat per kategori (t.ex. hur många "ja" som gissades rätt/fel och hur många "nej" som gissades rätt/fel).

b) Accuracy - andelen rätta gissningar delat på totala antalet gissningar.

c) Precision - av alla gånger modellen gissade "ja", hur stor andel var faktiskt rätt.

d) Recall - av alla som faktiskt var "ja", hur många lyckades modellen hitta.

e) F1-score - ett balanserat värde mellan precision och recall, bra när man vill ha en helhetsbild istället för att bara kolla på ett av måtten.

f) ROC-kurvan - en graf som visar hur bra modellen är på att skilja mellan klasserna vid olika tröskelvärden.

### 4. Vad är precision-recall tradeoff för något?

Precision-recall tradeoff betyder att man måste välja mellan hög precision eller hög recall, man kan sällan få båda maximerade samtidigt. Höjer man tröskeln för att räknas som "ja" blir modellen mer försiktig, precisionen går upp men recall går ner (missar fler). Sänker man tröskeln fångar modellen fler "ja" (högre recall) men precisionen sjunker eftersom fler blir felgissade.

### 5. Förklara översiktligt hur respektive modell fungerar: a) Logistisk regression, b) SVM, c) Beslutsträd, d) Ensemble learning, e) Random forest, f) Extra trees

a) Logistisk regression - räknar ut en chans i procent för att nåt tillhör en klass, är chansen över 50% blir svaret "ja", annars "nej".

b) SVM (support vector machines) - ritar en rak gränslinje med så bred säkerhetsmarginal som möjligt mellan grupperna.

c) Beslutsträd - sorterar in svaret genom enkla ja/nej-frågor tills den kommer fram till en klass.

d) Ensemble learning - flera olika modeller slås ihop och röstar om vilket svar som vinner.

e) Random forest - en hel skog av många olika beslutsträd som röstar tillsammans, ger säkrare resultat än ett enskilt träd.

f) Extra trees - liknar random forest men splittarna väljs ännu mer slumpmässigt istället för optimalt, vilket gör den snabbare och minskar risken för överanpassning, låter lite ologiskt men funkar bra i praktiken.

### 6. Vad innebär det att vi kan kolla på feature importance med hjälp av trädmodeller såsom beslutsträd eller random forest?

Det betyder att trädmodeller kan visa vilka variabler som hade störst betydelse för att gissa rätt, alltså hur mycket varje variabel påverkade modellens beslut. Variabler med hög feature importance har haft stor påverkan på resultatet, medan variabler med låg importance haft liten eller ingen påverkan.

### 11. Förklara vad classification_report-koden gör och tolka resultatet.

```python
from sklearn.metrics import classification_report

y_true = [0, 1, 2, 2, 2]
y_pred = [0, 0, 2, 2, 1]
target_names = ['class 0', 'class 1', 'class 2']
print(classification_report(y_true, y_pred, target_names=target_names))
```

classification_report visar precision, recall och f1 för varje klass, samt accuracy totalt. Support är hur många av varje klass som fanns i y_true.

Class 1 fick 0 på både precision och recall, den gissades aldrig rätt. Class 2 gick bäst. Totalt blev accuracy 0.60.

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report

df = pd.read_excel("data/hr_employee_data.xlsx")
df.head()
```

#### Bearbeta datan

Emp_Id är bara ett ID, den droppar vi. Salary är ordinal (low < medium < high) så vi ordinal-kodar den. Department är nominal så vi one-hot-kodar den.

```python
df = df.drop(columns="Emp_Id")

salary_mapping = {"low": 1, "medium": 2, "high": 3}
df["salary"] = df["salary"].map(salary_mapping)

df = pd.get_dummies(df, columns=["Department"], dtype=int)
df.head()
```

#### Dela upp i X och y, samt train/validation/test

```python
X = df.drop(columns="left")
y = df["left"]

X_train_full, X_test, y_train_full, y_test = train_test_split(X, y, test_size=0.2, random_state=40)
X_train, X_val, y_train, y_val = train_test_split(X_train_full, y_train_full, test_size=0.3, random_state=36)

print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

#### Träna två modeller

```python
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X_train, y_train)

forest = RandomForestClassifier(random_state=42)
hyperparams = {"max_depth": [5, 10, None]}

forest_gs = GridSearchCV(forest, hyperparams, scoring="accuracy", cv=5)
forest_gs.fit(X_train, y_train)

print(forest_gs.best_params_)
```

#### Utvärdera på valideringsdatan

```python
log_reg_pred = log_reg.predict(X_val)
forest_pred = forest_gs.predict(X_val)

print("Logistic Regression:")
print(classification_report(y_val, log_reg_pred))

print("Random Forest:")
print(classification_report(y_val, forest_pred))
```

#### Slutgiltig utvärdering på testdata

Random Forest vann på valideringsdatan, tränar om den på train+val innan vi testar den.

```python
best_model = RandomForestClassifier(**forest_gs.best_params_, random_state=42)
best_model.fit(X_train_full, y_train_full)

test_pred = best_model.predict(X_test)
print(classification_report(y_test, test_pred))
```

#### Träna om på hela datasetet och spara modellen

```python
import joblib

best_model.fit(X, y)  # tränar om på hela datasetet innan modellen sparas/produktionssätts

joblib.dump(best_model, "left_modell.pkl")
print("Modell sparad!")
```

Random forest var mycket bättre än logistisk regression, testresultatet blev 99% accuracy. Bra modell för att gissa vilka som kommer sluta.
