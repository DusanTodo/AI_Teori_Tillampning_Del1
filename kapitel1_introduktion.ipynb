### 3. Förklara följande:

#### a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och testdata?

Syftet med att dela upp data i träning, validering och test är att i slutändan få ett predikterat resultat som avviker så lite som möjligt från verkligt data. Träningsdata tränar olika modeller och måste vara störst. Valideringsdatan validerar resultat från träningsdatan och visar vilken modell som ger bäst resultat. Testdatan används sist för att testa den valda modellen på osedd data, så att man får ett ärligt resultat på hur modellen faktiskt presterar.

#### b) Vad är k-delad korsvalidering för något?

K-delad korsvalidering är en metod för att utvärdera hur bra en ML-modell presterar. Datan delas i k lika stora delar, t.ex. om k=5 blir det 5 delar (A, B, C, D, E). I varje omgång tränas modellen på 4 av delarna och valideras på den femte. Detta upprepas tills alla delar har fått vara valideringsdel en gång, så man får ut k stycken resultat. Till slut räknar man ut medelvärdet av de k resultaten, vilket ger ett mer rättvist och stabilt mått på modellens prestanda än om man bara gjort en enda train/validation-split.

#### c) Vad är root mean squared error (RMSE) för något?

RMSE (root mean squared error) är ett utvärderingsmått för regressionsmodeller. Det mäter felen, alltså hur mycket modellens gissningar skiljer sig från de verkliga värdena. Skillnaden kvadreras för att få alltid positiva värden, sedan räknar man medelvärdet av dessa kvadrerade fel, och till sist drar man kvadratroten för att få tillbaka felet i samma enhet som ursprungsdatan.

#### d) Vad är en hyperparameter för något? Vad är en parameter för något?

Hyperparameter är en inställning som vi bestämmer innan modellträningen börjar, den svarar på frågan hur vi lärt oss. Parameter är ett värde som modellen själv räknar ut och lär sig automatiskt från datan, den svarar på frågan vad har vi lärt oss.

#### e) Vad är grid search och hur hänger namnet "grid" och "search" ihop med processen? Vad innebär refit=True?

Grid search är en automatiserad metod för att hitta de bästa hyperparametrarna till en maskininlärningsmodell. Ordet "grid" syftar på ett rutnät/matris av olika hyperparametrar med värden, där varje korspunkt testas av algoritmen. Ordet "search" betyder att processen letar systematiskt igenom en förbestämd lista av olika värden. Att refit=True (standardvärdet i scikit-learns GridSearchCV) betyder att när sökningen är klar och den bästa kombinationen av hyperparametrar hittats, så tränas modellen om med de vinnande inställningarna.

#### f) Vad är kategorisk data och hur hanteras det? (nominal data, ordinal data, one-hot-encoding, dummy-variable-encoding, ordinal encoding)

Kategorisk data är icke-numerisk information från t.ex. ett dataset som kan delas in i flera kategorier, såsom färg, storlek, kön m.m. Eftersom ML-modeller bara förstår siffror måste den kategoriska datan omvandlas till numeriskt format.

Man skiljer på nominal data och ordinal data. Nominal data saknar rangordning, t.ex. färger. Ordinal data har en tydlig ordning, t.ex. betyg (låg, medel, hög).

För ordinal data använder man oftast ordinal encoding, där man rangordnar värdena, t.ex. låg=1, medel=2, hög=3.

För nominal data används one-hot-encoding eller dummy-variable-encoding. Vid one-hot-encoding görs varje kategori om till en egen kolumn med 1 eller 0. Vid dummy-variable-encoding skapar man k-1 kolumner istället för k (en kategori utelämnas). Dummy-variable-encoding används mest när man tränar linjära regressioner.

#### g) Vad är feature engineering för något?

Feature engineering handlar om att förbättra den indata som ska användas för träning och validering. Det delas in i tre delar: variabelselektion, att skapa nya variabler och transformering av variabler.

#### h) Vad menas med principle of parsimony?

Principle of parsimony kallas också Ockhams rakkniv. Det handlar om att när man har två liknande presterande modeller så väljer man den enklare, eftersom en mer komplicerad modell kan vara överanpassad (overfitted).
