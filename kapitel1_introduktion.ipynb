### 3. Förklara följande:

#### a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och testdata?

Syftet med att dela upp data i träning, validering och test är att i slutändan få ett predikterat resultat som avviker så lite som möjligt från verkligt data. Träningsdata tränar olika modeller och måste vara störst. Valideringsdatan validerar resultat från träningsdatan och visar vilken modell som ger bäst resultat. Testdatan används sist för att testa den valda modellen på osedd data, så att man får ett ärligt resultat på hur modellen faktiskt presterar.

#### b) Vad är k-delad korsvalidering för något?

K-delad korsvalidering är en metod för att utvärdera hur bra en ML-modell presterar. Datan delas i k lika stora delar, t.ex. om k=5 blir det 5 delar (A, B, C, D, E). I varje omgång tränas modellen på 4 av delarna och valideras på den femte. Detta upprepas tills alla delar har fått vara valideringsdel en gång, så man får ut k stycken resultat. Till slut räknar man ut medelvärdet av de k resultaten, vilket ger ett mer rättvist och stabilt mått på modellens prestanda än om man bara gjort en enda train/validation-split.

#### c) Vad är root mean squared error (RMSE) för något?

RMSE (root mean squared error) är ett utvärderingsmått för regressionsmodeller. Det mäter felen, alltså hur mycket modellens gissningar skiljer sig från de verkliga värdena. Skillnaden kvadreras för att få alltid positiva värden, sedan räknar man medelvärdet av dessa kvadrerade fel, och till sist drar man kvadratroten för att få tillbaka felet i samma enhet som ursprungsdatan.

#### d) Vad är en hyperparameter för något? Vad är en parameter för något?

Hyperparameter är en inställning som vi bestämmer innan modellträningen börjar, den svarar på frågan hur vi lärt oss. Parameter är ett värde som modellen själv räknar ut och lär sig automatiskt från datan, den svarar på frågan vad har vi lärt oss.
